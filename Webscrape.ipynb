```python
headers = {'User-Agent': 'Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/47.0.2526.106 Safari/537.36'}
# append df to a list
df = pd.DataFrame()
for season_id in range(2002, 2023):
    premurl = f"https://www.transfermarkt.co.uk/premier-league/tabelle/wettbewerb/GB1?saison_id={season_id}"
    laligaurl = f"https://www.transfermarkt.co.uk/laliga/tabelle/wettbewerb/ES1?saison_id={season_id}"
    bundesligaurl = f"https://www.transfermarkt.co.uk/bundesliga/tabelle/wettbewerb/L1?saison_id={season_id}"
    serieaurl = f"https://www.transfermarkt.co.uk/serie-a/tabelle/wettbewerb/IT1?saison_id={season_id}"
    ligue1url = f"https://www.transfermarkt.co.uk/ligue-1/tabelle/wettbewerb/FR1?saison_id={season_id}"
    urlList = [premurl,laligaurl,bundesligaurl,serieaurl,ligue1url]
    for each in urlList:
        response = requests.get(each, headers=headers)
        # do something with the response
        pageSoup = BeautifulSoup(response.content, 'html.parser')
        table = pageSoup.find('table', {'class': 'items'})
        rows = table.findAll('tr')
        data = []
        for row in rows:
            cols = row.findAll('td')
            cols = [col.text.strip() for col in cols]
            data.append(cols)
        df = pd.DataFrame(data)
        df = df.drop(df.columns[1], axis=1)
        df = df.iloc[1:]
        df = df.rename(columns={0:'Position', 2:'Club', 3:'Games', 4:'Wins', 5:'Draws', 6:'Losses', 7:'Goals', 8: 'GD', 9:'Points'})
        if season_id == 2009:
            df["Season"] = f'0{season_id-2000}/{season_id-1999}'
            print(f'0{season_id-2000}/{season_id-1999}')
        elif season_id < 2009:
            df["Season"] = f'0{season_id-2000}/0{season_id-1999}'
            print(f'0{season_id-2000}/0{season_id-1999}')
        else:
            df["Season"] = f'{season_id-2000}/{season_id-1999}'
            print(f'{season_id-2000}/{season_id-1999}')
        df['League'] = "Ligue1"
        df.to_csv('league1.csv', index=False, mode='a', header=False)
```
