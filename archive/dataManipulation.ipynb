```python
import pandas as pd
import plotly_express as px
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/Zohaib-A/footballData/main/TransferExpenditure.csv')
print(df)
```

```python
df['Club'].nunique()
```

```python
grouped = df.groupby(['League', 'Season'])['Expenditure'].sum().reset_index()
grouped
```

```python
topClubExpenditure = (
    df.groupby(
        by=["Club"]).sum()
    ).sort_values(by=['Balance'],ascending=True).head(20)

topClubExpenditure
```

```python
df2 = pd.read_csv('resources/league.csv')
df2
```

```python
merged_df = pd.merge(df, df2, on=['Club', 'Season', 'League'])
merged_df
```

```python
merged_df.loc[df['Club'] == 'Chelsea FC']
```

```python
total_spending = int(merged_df["Expenditure"].sum())
total_spending
```

```python
merged_df[['GF', 'GA']] = merged_df['Goals'].str.split(':', expand=True)
merged_df[['GF', 'GA']] = merged_df[['GF', 'GA']].astype(int)
merged_df
```

```python
clubResultsByYear = (
        merged_df.groupby(['Club', 'Season'])['Wins', 'Draws', 'Losses','Points', 'Expenditure'].sum().reset_index()
    )
clubResultsByYear.sort_values(by=['Points'],ascending=False).head(20)
```

```python
merged_df[merged_df['Position'] == 1].sort_values(by="Season", ascending=True)
```

```python
(merged_df.groupby(by=["Club","League"]).sum())
```

```python
x = merged_df.groupby(['Club', 'Season', 'Position'])[
        'Expenditure'].sum().reset_index()
x = x[x['Position'] == 9]
x = x.sort_values(
        by="Season", ascending=True)
x
```
